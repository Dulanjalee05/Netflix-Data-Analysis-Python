```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the cleaned dataset exported in Task 1
df = pd.read_csv('Cleaned_Netflix_Dataset.csv')
print("Dataset successfully loaded! Total rows:", len(df))
```

```python
# Calculate total counts for each content type
content_counts = df['type'].value_counts()

print("--- Total Content Counts ---")
print(content_counts)
```

```python
# Set visual style
sns.set_theme(style="whitegrid")

plt.figure(figsize=(12, 5))

# Bar Chart
plt.subplot(1, 2, 1)
sns.barplot(x=content_counts.index, y=content_counts.values, palette='Set2')
plt.title('Content Distribution (Bar Chart)')
plt.xlabel('Content Type')
plt.ylabel('Count')

# Pie Chart
plt.subplot(1, 2, 2)
plt.pie(content_counts, labels=content_counts.index, autopct='%1.1f%%', colors=['#66b3ff','#ff9999'], startangle=90)
plt.title('Content Proportions (Pie Chart)')

plt.tight_layout()
plt.show()
```

```python
# Calculate exact proportions and percentages
total_content = len(df)
movie_count = content_counts.get('Movie', 0)
tv_count = content_counts.get('TV Show', 0)

movie_pct = (movie_count / total_content) * 100
tv_pct = (tv_count / total_content) * 100

print("--- Content Proportions ---")
print(f"Total Titles : {total_content}")
print(f"Movies       : {movie_count} ({movie_pct:.2f}%)")
print(f"TV Shows     : {tv_count} ({tv_pct:.2f}%)")
```

```python
print("--- SUMMARY OF KEY FINDINGS ---")
print("1. Movies dominate the Netflix platform, representing ~69.7% of all titles.")
print("2. TV Shows account for the remaining ~30.3% of total available content.")
print("3. There are more than twice as many Movies (6,126) as TV Shows (2,664) in the dataset.")
```
